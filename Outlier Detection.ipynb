```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    'Employee_Salary': [
        20000, 22000, 25000, 28000, 30000,
        32000, 35000, 38000, 40000, 45000,
        50000, 55000, 60000, 65000, 200000
    ]
}

df = pd.DataFrame(data)

print(df)
```

```python
plt.figure(figsize=(8, 5))

plt.boxplot(df['Employee_Salary'])

plt.title('Employee Salary - Outlier Detection')
plt.ylabel('Salary')

plt.show()
```

```python
Q1 = df['Employee_Salary'].quantile(0.25)
Q3 = df['Employee_Salary'].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = df[
    (df['Employee_Salary'] < lower_limit) |
    (df['Employee_Salary'] > upper_limit)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)

print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)

print("\nPotential Outliers:")
print(outliers)
```

Findings
- A box plot was created to visualize employee salary distribution.
- The Interquartile Range (IQR) method was used to identify potential outliers.
- The salary value of 200000 is identified as a potential outlier in this sample dataset.
- Outliers should be investigated before deciding whether to remove them.
